```python
#part 1: encapsulation
%%script scala

class BankAccount(private var balance: Double){
  def getBalance(): Double = balance
  def deposit(amount: Double): Unit = {
    if(amount > 0) balance += amount
    else println("Invalid deposit amount")
  }
  def withdraw(amount: Double): Unit = {
    if(amount > 0 && amount <= balance) balance -= amount
    else println("Invalid withdrawal amount")
    }
}

object Encapsulation{
  def main(args: Array[String]): Unit = {
    val account = new BankAccount(1000)
    account.deposit(500)
    account.withdraw(200)
    println("Initial balance: " + account.getBalance)
}
}

Encapsulation.main(Array())
```

```python
%%script scala

#part 2: inheritance & overriding

#parent class
class Person(val name: String, val age: Int){
    def showInfo(): Unit = println(s"Name: name, Age: age")
}

class Employee(name: String, age: Int, val role: String)extends Person(name, age){
    override def showInfo(): Unit = println(s"Employee -> Name: $name, Age: $age, Role: $role")
}

object inheritanceDemo{
    def main(args: Array[String]): Unit = {
        val employee = new Employee("Alina", 30, "Developer")
        employee.showInfo()
    }
}

inheritanceDemo.main(Array())
```

```python
#abstraction

%%script scala

abstract class Shape{
    def area(): Double
}

class Circle(val radius: Double)extends Shape{
   def area(): Double = Math.PI * radius * radius
}

class Rectangle(val length: Double, val width: Double)extends Shape{
  def area(): Double = length * width
}

object AbstractionDemo {
  def main(args: Array[String]): Unit = {
    val shapes: List[Shape] = List(new Circle(5), new Rectangle(4, 6))
    shapes.foreach(s => println(s"Area: ${s.area()}"))
  }
}
AbstractionDemo.main(Array())
```

```python
#part 4 Ploymorphism

%%script scala

class Calculator{
  def add(a: Int, b: Int): Int = a + b
  def add(a: Double, b: Double): Double = a + b
}

class AdvancedCalculator extends Calculator{
  override def add(a: Int, b: Int): Int =
  {
    println("Using advancedcalculator logic")
    a + b + 10
}
}

object PolymorphismDemo {
  def main(args: Array[String]): Unit = {
    val calc = new Calculator
    println("Int Add: " + calc.add(3, 7))
    println("Double Add: " + calc.add(3.5, 2.5))

    val advCalc: Calculator = new AdvancedCalculator
    println("Overridden Add: " + advCalc.add(3, 7))
  }
}

PolymorphismDemo.main(Array())
```

```python
#part 5 traits

%%script scala

trait Printable{
  def printInfo(): Unit
}

trait Identifiable{
  def id: String
}

class Student(val id: String, val name: String) extends Printable with Identifiable{
  def printInfo(): Unit = println(s"Student -> ID: $id, Name: $name")
}

object TraitDemo {
  def main(args: Array[String]): Unit = {
    val student = new Student("S101", "Alina")
    student.printInfo()
  }
}

TraitDemo.main(Array())
```
